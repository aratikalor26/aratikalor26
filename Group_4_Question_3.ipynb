```python
def average_height(url_list):
    # list to store heights
    heights = []
    names = []

    # header source : https://www.zenrows.com/blog/web-scraping-headers#accept-language
    headers = {
        'User-Agent': 'Mozilla/5.0 (Linux; Android 11; SAMSUNG SM-G973U) AppleWebKit/537.36 (KHTML, like Gecko) SamsungBrowser/14.2 Chrome/87.0.4280.141 Mobile Safari/537.36'
    }

    # visit each url in the list
    for url in url_list:
        url = url.strip()  # remove spaces

        # making a request to the server
        page = requests.get(url, headers=headers)

        # scrape data only if connection is successful
        if page.status_code == 200:
            soup = BeautifulSoup(page.content, 'html.parser')

            # find all td tags with class of height
            raw_heights = soup.find_all('td', class_='height')

            # find all td tags with a class of sidearm-table-player-name
            raw_names = soup.find_all('td', class_='sidearm-table-player-name')

            # extracting the name from the name tags
            for name in raw_names:
                names.append(name.text.strip())

            # extracting the raw height from the height tags
            for height in raw_heights:
                height_text = height.text.strip()
                if '-' in height_text:
                    try:
                        # splitting the string by the '-'
                        feet, inches = height_text.split('-')
                        # converting height to inches
                        total_inches = int(feet) * 12 + int(inches)
                        heights.append(total_inches)
                    except:
                        continue  # skip if conversion fails

    # organizing the data as a dictionary
    min_len = min(len(names), len(heights))

    # handle empty data case
    if min_len == 0:
        return pd.DataFrame(columns=["Name", "Height_in"])

    data = {"Name": names[:min_len], "Height_in": heights[:min_len]}

    # creating a dataframe from the dictionary
    df = pd.DataFrame(data)
    return df
```

```python
#Q3 Part 1. Scrape data and compile a dataframe of all the names and heights of the players on the men’s swimming teams. Output this data as a csv file.
Mens_Swimming_and_Diving_df = average_height(sports_teams["Mens_Swimming_and_Diving"])
Mens_Swimming_and_Diving_df

#save to csv file
Mens_Swimming_and_Diving_df.to_csv("mens_swimming_heights.csv", index=False)
```
